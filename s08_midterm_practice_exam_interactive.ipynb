### Question 1 (Multiple Choice)
What is the output of the following code?
```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return f"{self.name} makes a sound"

class Dog(Animal):
    def speak(self):
        return f"{self.name} barks"

pet = Dog("Max")
print(pet.speak())
```

a) `Max makes a sound`  
b) `Max barks`  
c) `Dog barks`  
d) Error: speak() is defined twice

````python
# ANSWER: b) Max barks
# Explanation: The Dog class overrides the speak() method from Animal class.
# When we call pet.speak(), it uses the Dog's version of the method.

# Test it yourself:
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

class Dog(Animal):
    def speak(self):
        return f"{self.name} barks"

pet = Dog("Max")
print(pet.speak())
````

### Question 3 (Short Answer)
Create a Python class called `BankAccount` with:
- An `__init__` method that accepts `account_number` and `balance` (default to 0)
- A method called `deposit(amount)` that adds to the balance and returns the new balance
- A method called `withdraw(amount)` that subtracts from balance if funds are sufficient, returns True if successful, False otherwise

````python
# ANSWER:
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        return False

# Test the class:
account = BankAccount("12345", 100)
print(f"Initial balance: ${account.balance}")
print(f"After deposit of $50: ${account.deposit(50)}")
print(f"Withdraw $75 successful? {account.withdraw(75)}")
print(f"Current balance: ${account.balance}")
print(f"Withdraw $100 successful? {account.withdraw(100)}")
print(f"Final balance: ${account.balance}")
````

### Question 4 (Short Answer)
What is method overriding in inheritance? Provide a brief explanation and a simple example.

````python
# ANSWER:
# Method overriding occurs when a child class provides its own implementation
# of a method that already exists in the parent class. The child's version
# replaces the parent's version when called on child objects.

# Example:
class Vehicle:
    def start(self):
        return "Vehicle starting"

class Car(Vehicle):
    def start(self):  # This overrides the parent's start method
        return "Car engine starting"

class Bicycle(Vehicle):
    def start(self):  # This also overrides the parent's start method
        return "Bicycle ready to pedal"

# Test the overriding:
vehicle = Vehicle()
car = Car()
bike = Bicycle()

print(vehicle.start())  # Uses Vehicle's method
print(car.start())      # Uses Car's overridden method
print(bike.start())     # Uses Bicycle's overridden method
````

Which of the following demonstrates that functions are first-class objects in Python?

a) Functions can only be called with parentheses  
b) Functions can be assigned to variables and passed as arguments  
c) Functions must have a return statement  
d) Functions can only accept primitive data types

````python
# ANSWER: b) Functions can be assigned to variables and passed as arguments

# Demonstration of first-class functions:
def greet(name):
    return f"Hello, {name}!"

def shout(name):
    return f"HELLO, {name.upper()}!"

# Assign function to variable
my_func = greet
print(my_func("Alice"))

# Pass function as argument
def apply_greeting(func, name):
    return func(name)

print(apply_greeting(greet, "Bob"))
print(apply_greeting(shout, "Charlie"))

# Store functions in a list
functions = [greet, shout]
for f in functions:
    print(f("David"))
````

### Question 11 (Multiple Choice)
What will be the output of this code?
```python
def outer(x):
    def inner():
        return x * 2
    return inner

func = outer(5)
print(func())
```

a) 5  
b) 10  
c) Error: inner is not defined  
d) None

````python
# ANSWER: b) 10

# Explanation and demonstration:
def outer(x):
    def inner():
        return x * 2
    return inner

func = outer(5)
print(f"Result: {func()}")

print("""
Explanation:
1. outer(5) creates an inner function that has access to x=5 (closure)
2. outer returns the inner function (not calling it)
3. func now holds the inner function
4. func() calls inner(), which returns x * 2 = 5 * 2 = 10

This demonstrates closures - inner functions can access variables from outer scope.
""")
````

### Question 12 (Short Answer)
Write a list comprehension that takes a list of strings and returns a list containing only the strings that are longer than 3 characters, converted to uppercase.

Example input: `["hi", "hello", "a", "world", "ok"]`  
Expected output: `["HELLO", "WORLD"]`

````python
# ANSWER:
strings = ["hi", "hello", "a", "world", "ok"]

# List comprehension solution:
result = [s.upper() for s in strings if len(s) > 3]
print(f"Result: {result}")

# Breaking it down:
print("""
Explanation of: [s.upper() for s in strings if len(s) > 3]

1. 'for s in strings' - iterate through each string
2. 'if len(s) > 3' - filter condition (only strings longer than 3 chars)
3. 's.upper()' - transformation (convert to uppercase)

Equivalent using regular loop:
""")

# Equivalent with regular loop:
result_loop = []
for s in strings:
    if len(s) > 3:
        result_loop.append(s.upper())
print(f"Loop result: {result_loop}")
````

### Question 14 (Short Answer)

### Question 15 (Multiple Choice)
What is the purpose of using `with open('file.txt') as f:` instead of just `open('file.txt')`?

a) It makes the code run faster  
b) It automatically closes the file when done, even if an error occurs  
c) It allows writing to read-only files  
d) It prevents other programs from accessing the file

````python
# ANSWER: b) It automatically closes the file when done, even if an error occurs

print("""
Explanation: The 'with' statement is a context manager that ensures proper
resource cleanup. For files, it guarantees the file will be closed even if
an error occurs.

Comparison:
""")

# Without 'with' - risky
print("""
# Without 'with' (NOT recommended):
file = open('file.txt')
data = file.read()  # If this raises an error...
file.close()        # ...this never runs! File stays open!
""")

# With 'with' - safe
print("""
# With 'with' (recommended):
with open('file.txt') as file:
    data = file.read()  # Even if this raises an error...
# File is automatically closed here, no matter what!
""")

# Demonstration with actual file
import tempfile
import os

# Create a temporary file for demonstration
with tempfile.NamedTemporaryFile(mode='w', delete=False, suffix='.txt') as temp:
    temp.write("Hello, World!")
    temp_name = temp.name

# Read using context manager
with open(temp_name, 'r') as f:
    print(f"File content: {f.read()}")
    print(f"File closed inside 'with'? {f.closed}")

print(f"File closed outside 'with'? {f.closed}")

# Clean up
os.unlink(temp_name)
````

### Question 16 (Multiple Choice)
What is the main difference between a list comprehension and a generator expression?

a) Generator expressions use square brackets, list comprehensions use parentheses  
b) List comprehensions create the entire list in memory, generator expressions produce items on demand  
c) Generator expressions are faster for all operations  
d) There is no difference, they are interchangeable

````python
# ANSWER: b) List comprehensions create the entire list in memory, generator expressions produce items on demand

import sys

# List comprehension - uses square brackets []
list_comp = [x**2 for x in range(1000)]
print(f"List comprehension type: {type(list_comp)}")
print(f"List comprehension size: {sys.getsizeof(list_comp)} bytes")
print(f"First 5 elements: {list_comp[:5]}")

print()

# Generator expression - uses parentheses ()
gen_expr = (x**2 for x in range(1000))
print(f"Generator expression type: {type(gen_expr)}")
print(f"Generator expression size: {sys.getsizeof(gen_expr)} bytes")
print(f"First 5 elements: {[next(gen_expr) for _ in range(5)]}")

print("""
Key differences:
1. Memory: List stores all values, generator produces on-demand
2. Speed: Generator is faster to create, list is faster to access repeatedly
3. Use cases:
   - List: When you need all values multiple times
   - Generator: When processing large data or one-time iteration
""")
````

### Question 17 (Short Answer)
Write a generator function that yields the first `n` even numbers.

Example: `get_evens(5)` should yield: 0, 2, 4, 6, 8

````python
# ANSWER:
def get_evens(n):
    count = 0
    num = 0
    while count < n:
        yield num
        num += 2
        count += 1

# Test the generator
print("Using the generator:")
for even in get_evens(5):
    print(even, end=" ")
print()

# Alternative implementation:
def get_evens_v2(n):
    for i in range(n):
        yield i * 2

print("\nAlternative version:")
print(list(get_evens_v2(5)))

print("""
Explanation:
- 'yield' makes this a generator function
- Each call to the generator returns the next value
- State is preserved between calls
- More memory efficient than creating a full list
""")
````

### Question 18 (Short Answer)
Convert this nested loop into a single list comprehension:
```python
result = []
for i in range(3):
    for j in range(3):
        if i != j:
            result.append((i, j))
```

````python
# ANSWER:

# Original nested loop:
result_loop = []
for i in range(3):
    for j in range(3):
        if i != j:
            result_loop.append((i, j))
print(f"Original result: {result_loop}")

# List comprehension solution:
result = [(i, j) for i in range(3) for j in range(3) if i != j]
print(f"Comprehension result: {result}")

print("""
Explanation of: [(i, j) for i in range(3) for j in range(3) if i != j]

Reading order (same as nested loops):
1. for i in range(3)     - outer loop
2. for j in range(3)     - inner loop
3. if i != j             - condition
4. (i, j)                - what to append

This creates all pairs (i, j) where i and j are different.
""")

# Verify they're equal:
print(f"\nAre they equal? {result == result_loop}")
````

### Question 19 (Multiple Choice)
What will this generator function produce?
```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

gen = countdown(3)
print(next(gen))
print(next(gen))
```

a) 3, 2  
b) 2, 1  
c) 1, 2  
d) Error: generators don't support next()

````python
# ANSWER: a) 3, 2

def countdown(n):
    while n > 0:
        yield n
        n -= 1

gen = countdown(3)
print(f"First next(): {next(gen)}")
print(f"Second next(): {next(gen)}")
print(f"Third next(): {next(gen)}")

print("""
Explanation:
1. countdown(3) creates a generator starting at n=3
2. First next() yields 3, then decrements n to 2
3. Second next() yields 2, then decrements n to 1
4. Third next() would yield 1, then n becomes 0
5. Fourth next() would raise StopIteration (no more values)

Generators maintain state between next() calls!
""")

# Demonstrate StopIteration:
gen2 = countdown(1)
print(f"\nNew generator with n=1:")
print(f"First call: {next(gen2)}")
try:
    print(f"Second call: {next(gen2)}")
except StopIteration:
    print("StopIteration raised - generator exhausted!")
````
