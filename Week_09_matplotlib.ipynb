```python
data_a = [1, 2, 3, 4]
data_b = [1, 4, 9, 16]
data_c = [1, 8, 27, 64]
fig, ax = plt.subplots(2, 2)
ax[0, 0].plot(data_a, data_b)
ax[1, 1].plot(data_b, data_c)
ax[0, 1].plot(data_b, data_a)
ax[1, 0].plot(data_c, data_a)
fig.savefig('subplots_are_awesome.png')
```

```python
import numpy as np
```

```python
x = np.arange(0, 100)
y = x + 10*(np.random.rand(x.shape[0])) # add noise

fig, ax = plt.subplots(1, 1)
ax.plot(x, y, linestyle='--', color='C1', label='data')
ax.grid(True)
ax.set_title('Babby\'s first line plot')
ax.set_xlabel('x data')
ax.set_ylabel('y data')
ax.legend(loc='lower right')
fig.savefig('TAC 216 first plot.png')
```

```python
x = np.arange(0, 10)
y = 2 * x + 6
fig, ax = plt.subplots(1, 1)
ax.plot(x, y)
fig.savefig('linear.png')
```

```python
x = np.arange(0, 2*np.pi, 0.1)
y = np.sin(x)
fig, ax = plt.subplots(1, 1)
ax.plot(x, y)
ax.set(xlabel='x', ylabel='y', title='Trig is fun!')
fig.savefig('sin.png')
```

```python
import pandas as pd
```

```python
# file_path = 'iris.csv'
df_iris = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/refs/heads/master/iris.csv")

fig, ax = plt.subplots(1, 3, squeeze=True, sharey=False)
ax[0].scatter(df_iris['petal_length'], df_iris['petal_width'])
ax[1].scatter(df_iris['sepal_length'], df_iris['petal_width'])
ax[2].scatter(df_iris['sepal_width'], df_iris['petal_width'])
fig.tight_layout()
fig.savefig('iris_petal_scatterplot.png')
```
